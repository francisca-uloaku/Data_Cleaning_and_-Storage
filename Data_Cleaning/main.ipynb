```python
#importing the data
import pandas as pd
import numpy as np

urban_dataset = pd.read_csv("UrbanNest_Retail_Dirty_Customer_Order_Dataset.csv")

print(urban_dataset.head())
```

```python
#Standardizing the customer column
urban_dataset["customer_name"] = urban_dataset["customer_name"].str.strip()
print(urban_dataset["customer_name"].head(10))
```

```python

urban_dataset["gender"] = urban_dataset["gender"].replace({
    "m": "male",
    "f": "female",
    "M": "male",
    "F": "female"})
print(urban_dataset["gender"].head(10))
```

```python
urban_dataset["unit_price"] = pd.to_numeric(urban_dataset["unit_price"], errors='coerce').fillna(0.0).astype(float)
print(urban_dataset["unit_price"])
```

```python
# Standardizing the date
urban_dataset["order_date"] = pd.to_datetime(
    urban_dataset["order_date"],
    errors="coerce")
print(urban_dataset["order_date"])
```

```python
#Confirming the data types
urban_dataset.dtypes
```

```python
#Confirming the data stored 
print(urban_dataset["delivery_days"])
```

```python
urban_dataset.isnull().sum()
```

```python
#Ensuring the data set follows set variations
gender_map = {'Male': 'Male', 'MALE': 'Male', 'M': 'Male', 'Female': 'Female', 'F': 'Female'}
urban_dataset['gender'] = urban_dataset['gender'].map(gender_map)
print(urban_dataset['gender'])
```

```python
#Standardizing missing values
urban_dataset = urban_dataset.replace(to_replace=[r'^\s*$', 'nan', 'NaN'], value=np.nan, regex=True)
print(urban_dataset['gender'])
```

```python
urban_dataset["delivery_fee"] = urban_dataset["delivery_fee"].fillna(0)
print(urban_dataset["delivery_fee"].head())
```

```python
print(urban_dataset.head())
```

```python
#Handling missing data
urban_dataset['gender'] = urban_dataset['gender'].fillna('Unknown')
urban_dataset['customer_segment'] = urban_dataset['customer_segment'].fillna('Unknown')
urban_dataset['order_status'] = urban_dataset['order_status'].fillna('Unknown')
print(urban_dataset.head())
```

```python
#Ensuring No column is left empty and Ensuring the data types are maintained
urban_dataset['payment_method'] = urban_dataset['payment_method'].astype(str).str.strip().str.title()
urban_dataset['payment_method'] = urban_dataset['payment_method'].replace({'Nan': 'Unknown', 'None': 'Unknown'})

urban_dataset['sales_channel'] = urban_dataset['sales_channel'].astype(str).str.strip().str.title()
urban_dataset['sales_channel'] = urban_dataset['sales_channel'].replace({'Nan': 'Unknown', 'None': 'Unknown'})


urban_dataset['delivery_status'] = urban_dataset['delivery_status'].astype(str).str.strip().str.title()
urban_dataset['delivery_status'] = urban_dataset['delivery_status'].replace({'Nan': 'Unknown', 'None': 'Unknown'})

urban_dataset['order_status'] = urban_dataset['order_status'].astype(str).str.strip().str.title()
urban_dataset['order_status'] = urban_dataset['order_status'].replace({'Nan': 'Unknown', 'None': 'Unknown'})


urban_dataset['return_flag'] = urban_dataset['return_flag'].astype(str).str.strip().str.title()
urban_dataset['return_flag'] = urban_dataset['return_flag'].replace({'Nan': 'Unknown', 'None': 'Unknown'})

urban_dataset['delivery_days'] = pd.to_numeric(urban_dataset['delivery_days'], errors='coerce').fillna(0).astype(int)
urban_dataset['delivery_fee'] = pd.to_numeric(urban_dataset['delivery_fee'], errors='coerce').fillna(0.0)

urban_dataset['age_group'] = urban_dataset['age_group'].astype(str).str.strip().str.title()
urban_dataset['age_group'] = urban_dataset['age_group'].replace({
    'Nan': 'Unknown', 
    'None': 'Unknown', 
    'Null': 'Unknown', 
    '': 'Unknown'
})

print(urban_dataset[['age_group', 'payment_method', 'sales_channel', 'order_status', 'delivery_status']].head(15))
```

```python
#Esuring that we do not have any duplicated column
urban_duplicate = urban_dataset[urban_dataset.duplicated()]
print(urban_duplicate)
print(urban_dataset.duplicated().sum())
```
